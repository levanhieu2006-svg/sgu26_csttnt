```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Khai báo 12 tên cột chuẩn của bộ dữ liệu Wine Quality
col_names = ['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar',
             'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density',
             'pH', 'sulphates', 'alcohol', 'quality']

# 2. Đọc file, báo cho nó biết là "file này không có tiêu đề đâu", và gán tên cột vào
df = pd.read_csv('winequality-red.csv', sep=',', header=None, names=col_names)

print("✅ ĐÃ GẮN TÊN CỘT THÀNH CÔNG! Danh sách cột hiện tại:")
print(df.columns.tolist())
print("-" * 50)

# Cài đặt phong cách biểu đồ
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 12})

# --- HÌNH 1: BIỂU ĐỒ PHÂN PHỐI CHẤT LƯỢNG RƯỢU ---
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='quality', palette='viridis')
plt.title('Phân phối Điểm chất lượng Rượu vang đỏ (Quality)', fontsize=14, fontweight='bold')
plt.xlabel('Điểm chất lượng (Quality)')
plt.ylabel('Số lượng mẫu')
plt.tight_layout()
plt.show()

# --- HÌNH 2: MA TRẬN TƯƠNG QUAN (CORRELATION HEATMAP) ---
plt.figure(figsize=(12, 9))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='RdBu_r', fmt='.2f', linewidths=0.5)
plt.title('Ma trận tương quan giữa các đặc trưng hóa lý', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# --- HÌNH 3: BIỂU ĐỒ HỘP (BOXPLOT) CHỨNG MINH SỰ CẦN THIẾT CỦA CHUẨN HÓA ---
plt.figure(figsize=(14, 6))
# Lấy tất cả các biến đầu vào để vẽ (bỏ cột quality ra)
input_features = [col for col in df.columns if col != 'quality']
sns.boxplot(data=df[input_features], palette='Set2')
plt.title('Sự chênh lệch thang đo giữa các đặc trưng (Trước khi chuẩn hóa)', fontsize=14, fontweight='bold')
plt.ylabel('Giá trị')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc lại dữ liệu và gắn tên cột cho chắc cú
col_names = ['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar',
             'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density',
             'pH', 'sulphates', 'alcohol', 'quality']
df = pd.read_csv('winequality-red.csv', sep=',', header=None, names=col_names)

sns.set_theme(style="whitegrid")

# --- HÌNH 4: BIỂU ĐỒ HỘP CHI TIẾT (ALCOHOL & VOLATILE ACIDITY vs QUALITY) ---
# Trực quan hóa xem cồn và axit ảnh hưởng tới điểm số như thế nào
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.boxplot(ax=axes[0], x='quality', y='alcohol', data=df, palette='Blues')
axes[0].set_title('Sự phân bố Nồng độ Cồn (Alcohol) theo Điểm chất lượng', fontweight='bold')
axes[0].set_xlabel('Điểm chất lượng (Quality)')
axes[0].set_ylabel('Nồng độ cồn (%)')

sns.boxplot(ax=axes[1], x='quality', y='volatile acidity', data=df, palette='Reds')
axes[1].set_title('Sự phân bố Axit bay hơi (Volatile Acidity) theo Điểm chất lượng', fontweight='bold')
axes[1].set_xlabel('Điểm chất lượng (Quality)')
axes[1].set_ylabel('Axit bay hơi (g/dm3)')

plt.tight_layout()
plt.show()
print("-> HÌNH 4: Biểu đồ hộp thể hiện sự biến thiên của 2 đặc trưng quan trọng nhất theo chất lượng rượu.\n")

# --- HÌNH 5: BIỂU ĐỒ PHÂN BỐ KÉP (PAIRPLOT) CÁC ĐẶC TRƯNG TOP ĐẦU ---
# Vẽ sự phân tách các lớp điểm cao/thấp dựa trên các biến mạnh nhất
plt.figure(figsize=(10, 10))
# Nhóm điểm chất lượng lại để dễ nhìn (Tệ: 3,4 | Trung bình: 5,6 | Tốt: 7,8)
df['quality_label'] = pd.cut(df['quality'], bins=[2, 4.5, 6.5, 9], labels=['Tệ', 'Trung bình', 'Tốt'])
cols_for_pairplot = ['alcohol', 'volatile acidity', 'sulphates', 'quality_label']

sns.pairplot(df[cols_for_pairplot], hue='quality_label', palette='Set1', diag_kind='kde', corner=True)
plt.suptitle('Biểu đồ Pairplot: Sự giao thoa (Overlap) giữa các nhóm chất lượng', y=1.02, fontweight='bold')
plt.show()
print("-> HÌNH 5: Biểu đồ này y hệt Hình 7 trong bài mẫu của bạn, dùng để chứng minh ranh giới giữa các nhóm rượu bị mờ nhạt (chồng lấn lên nhau).\n")

# --- HÌNH 6: BIỂU ĐỒ MẬT ĐỘ (KDE PLOT) CHỨNG MINH ĐỘ LỆCH CỦA DỮ LIỆU ---
plt.figure(figsize=(8, 5))
sns.kdeplot(df['residual sugar'], fill=True, color="purple", label="Residual Sugar")
sns.kdeplot(df['chlorides'], fill=True, color="orange", label="Chlorides")
plt.title('Phân phối đuôi dài (Long-tail) và Ngoại lệ của Đường dư & Muối', fontweight='bold')
plt.xlabel('Giá trị')
plt.ylabel('Mật độ')
plt.legend()
plt.tight_layout()
plt.show()
print("-> HÌNH 6: Biểu đồ này dùng để giải thích tại sao nhóm quyết định KHÔNG XÓA Outliers (ngoại lệ).")
```
